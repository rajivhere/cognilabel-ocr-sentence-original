```python
from bidi.algorithm import get_display

def normalize_label(txt: str, text_dir="ltr", label_order_policy="logical") -> str:
    txt = txt.strip()
    txt = " ".join(txt.split())

    if text_dir == "rtl":
        if label_order_policy == "logical":
            return txt
        if label_order_policy == "visual_bidi":
            return get_display(txt, base_dir="R")
        if label_order_policy == "reverse_chars":
            return txt[::-1]
        raise ValueError(f"Unknown label_order_policy={label_order_policy}")

    return txt
```

```python
from bidi.algorithm import get_display
import re

TRAILING_NEUTRALS_RE = re.compile(
    r"""([\s\.\,\،\؛\:\!\؟\?\…\"\'\)\]\}\»\”\“\%]+)$"""
)

def rtl_trailing_neutrals_only(txt: str) -> str:
    
    txt = " ".join(txt.strip().split())
    m = TRAILING_NEUTRALS_RE.search(txt)
    if not m:
        return txt

    tail = m.group(1).strip()
    body = txt[:m.start()].rstrip()

    if not tail:
        return txt

    return f"{tail} {body}"


def show_policy_outputs(samples):
    for s in samples:
        policies = {
            "logical": s,
            "reverse_chars": s[::-1],
            "bidi_visual_R": get_display(s, base_dir="R"),
            "trailing_neutrals": rtl_trailing_neutrals_only(s),
        }

        print("=" * 100)
        print("DB/logical:", repr(s))

        for name, out in policies.items():
            print("-" * 60)
            print(name)
            print("display:", out)
            print("repr:   ", repr(out))
            print("chars:  ", " | ".join(ch if ch != " " else "␠" for ch in out))


show_policy_outputs(samples)
```

```python
def normalize_label(txt: str, text_dir="ltr", label_order_policy="logical") -> str:
    txt = txt.strip()
    txt = " ".join(txt.split())

    if text_dir != "rtl":
        return txt

    if label_order_policy == "logical":
        return txt

    if label_order_policy == "reverse_chars":
        return txt[::-1]

    if label_order_policy == "bidi_visual":
        from bidi.algorithm import get_display
        return get_display(txt, base_dir="R")

    if label_order_policy == "rtl_trailing_neutrals_only":
        return rtl_trailing_neutrals_only(txt)

    raise ValueError(f"Unknown label_order_policy={label_order_policy}")
```

```python
from bidi.algorithm import get_display
import unicodedata

samples = [
    'من سلطانَ . ومع ذلك تمكن من الحصول علىٰ "255" نقشاً رجع بها إلىٰ أوروبه.',
    'و بالحضر و بالعرب" . و وردث في النص: "193"@ ملكادي عرب"',
    'حوالي 30 كيلومتراً إلى السمال الغربي من مدينة صبيا المعروفه .',
    '@ 542م  والذي عن دي إن العرب الجنوبين لم يفههوا هذا المعني من اللفظة الا بعد دخولهم فى',
]

for s in samples:
    print("=" * 100)
    print("LOGICAL / DB:")
    print(s)
    print("repr:", repr(s))

    print("\nNAIVE REVERSE:")
    print(s[::-1])
    print("repr:", repr(s[::-1]))

    print("\nBIDI VISUAL R:")
    b = get_display(s, base_dir="R")
    print(b)
    print("repr:", repr(b))
```

```python
def inspect_bidi(s):
    for i, ch in enumerate(s):
        visible = ch if ch != " " else "␠"
        print(
            f"{i:02d} {visible} "
            f"U+{ord(ch):04X} "
            f"name={unicodedata.name(ch, 'UNKNOWN')} "
            f"bidi={unicodedata.bidirectional(ch)}"
        )

inspect_bidi('@ 542م  والذي')
```

```python
import re

BOUNDARY_NEUTRALS_RE = re.compile(
    r"""([\s\.\,\،\؛\:\!\؟\?\…\"\'\)\]\}\»\”\“\%]+)$"""
)

def rtl_boundary_neutrals(txt: str) -> str:
    txt = " ".join(txt.strip().split())

    m = BOUNDARY_NEUTRALS_RE.search(txt)
    if not m:
        return txt

    tail = m.group(1).strip()
    body = txt[:m.start()].rstrip()

    if not tail:
        return txt

    return f"{tail} {body}"
```
